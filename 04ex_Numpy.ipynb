## 2\. Find the outer product of the following two vecotrs

```python
u = np.array([1,3,5,7])
v = np.array([2,4,6,8])
```

Do this in the following ways:

   * Using the function outer in numpy
   * Using a nested for loop or list comprehension
   * Using numpy broadcasting operatoins


````python
u = np.array([1,3,5,7])
v = np.array([2,4,6,8])
print('u =',u,'\nv =',v,'\n')
# 1. Outer function
print('Outer function\n',np.outer(u,v),'\n')

# 2. Nested loop / list comprehension
print('Nested loop/list comprehension\n',
      np.array([i*j for i in u for j in v]).reshape((len(u),len(v))),'\n')
# 3. Broadcasting operations
u_m = np.tile(u,(len(v),1)).T
print('Broadcastig\n', u_m*v)
````

## 4\. Use np.linspace to create an array of 100 numbers between 0 and 2π (includsive).

  * Extract every 10th element using slice notation
  * Reverse the array using slice notation
  * Extract elements where the absolute difference between the sine and cosine functions evaluated at that element is less than 0.1
  * Make a plot showing the sin and cos functions and indicate where they are close

### Solution

````python
a_pi = np.linspace(0,np.pi,100)
print(a_pi,'\n')

a_s = a_pi[::10]
print('Only the 10th elements:\n',a_s,'\n')

print('Reverse matrix:\n',a_pi[::-1],'\n')

a_sc = [i for i in a_pi if np.abs(np.sin(i)-np.cos(i))<0.1]
print('Absolute difference between sine and cosine less than 0.1:\n',a_sc)
````

````python
import matplotlib.pyplot as plt
plt.plot(a_pi,np.sin(a_pi),label='Sine')
plt.plot(a_pi,np.cos(a_pi),label='Cosine')
plt.plot(a_sc,np.sin(a_sc),'o',label='close values')
plt.plot(a_sc,np.cos(a_sc),'o')
plt.xlabel('angle (rad)')
plt.ylabel('sine, cosine')
plt.legend()
plt.show()
````

## 6\. Use broadcasting to create a grid of distances

Route 66 crosses the following cities in the US: Chicago, Springfield, Saint-Louis, Tulsa, Oklahoma City, Amarillo, Santa Fe, Albuquerque, Flagstaff, Los Angeles
The corresponding positions in miles are: 0, 198, 303, 736, 871, 1175, 1475, 1544, 1913, 2448

  * Construct a 2D grid of distances among each city along Route 66
  * Convert that in km (those savages...)

### Solution

````python
position = np.array([0, 198, 303, 736, 871, 1175, 1475, 1544, 1913, 2448])
print(position)
#Create a 2D grid with the (ecuclidean) distance based in the cities position 
pos_m = np.tile(position,(len(position),1)).T
print(pos_m)
dist = np.sqrt(pos_m**2 + position**2)
print(distance)
#Transform distance to km (1 mile --> 1.60934 km)
dist_km = dist*1,60934
print(dist_km)
````
