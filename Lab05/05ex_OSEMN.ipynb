```python
with open('credit_card.dat', mode='r') as file:
    for line in file:
        #print(line)
        char = [x for x in line] #string with the split digits
        idx=0 #used as an index
        cn = [] #for storing the card number
        while (idx<len(char)): 
            c = char[idx:idx+6] #reads 6 digits at a time
            c = ''.join(c) #join the 6 digits as one string
            cd = chr(int(c,2)) #convers the string to number and the number to the characters (ASCII code)
            cn.append(cd) #Save every character
            idx+=6 #increase the index
        print(''.join(cn)) #print the card number stored in cn list
```

```python
# Optimized version:
# 1. Strings are already iterable so I don't need the:
#     char = [x for x in line]
# 2. Each line has a '\n' so using .strip() is better, and also it's better
#    to remove the 4 last digits beforehand.
# 3. I can use a for loop with a step of 6 instead of the while loop.
with open('credit_card.dat', mode='r') as file:
    for line in file:
        line = line.strip()[:-4]  # Remove newline/whitespace and padding
        cn = []
        for idx in range(0, len(line), 6):
            c = line[idx:idx+6]
            cd = chr(int(c, 2))
            cn.append(cd)
        print(''.join(cn))
```
